```
import numpy as np
import matplotlib.pyplot as plt

# 1. seed 오타 수정 (sedd -> seed)
np.random.seed(2025)

N = 80
x = np.random.rand(N)
y = np.random.rand(N)

# 2. 할당 연산자 중복 제거
colors = np.random.rand(N)
area = (20 * np.random.rand(N)) ** 2

plt.scatter(x, y, s=area, c=colors, alpha=0.7)
plt.colorbar()
plt.show()
```

```
x = np.random.randint(-100,100,1000)
y = np.random.randint(-100,100,1000)

mask1 = abs(x) > 50
mask2 = abs(y) > 50
x = x[mask1 + mask2]
y = y[mask1 + mask2]

plt.scatter(x, y,c=x,cmap='jet', alpha=0.7)
plt.colorbar()
plt.show()
```

```
import matplotlib.pyplot as plt

# 1. 데이터 준비
# 1이 3개, 2가 2개, 3이 1개, 4가 1개, 5가 1개, 6이 4개 있는 리스트입니다.
data = [1, 1, 1, 2, 2, 3, 4, 5, 6, 6, 6, 6]

# 2. 그림 크기 지정
plt.figure(figsize=(6, 4))

# 3. 히스토그램 그리기
# rwidth=0.98: 막대 간의 간격을 아주 살짝 띄워서 구분하기 쉽게 함
# bins=[.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5]:
# 각 숫자가 막대의 정중앙에 오도록 구간의 경계(담장)를 설정함
plt.hist(data, rwidth=0.98, bins=[.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5])

# 4. X축 눈금 설정 (1부터 6까지 표시)
plt.xticks(range(1, 7))

# 5. 그래프 출력
plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np

# 1. 조작된 주사위 데이터 생성
# 1부터 6까지의 숫자 중 하나를 선택 (1,000,000번 반복)
# p는 각 숫자가 나올 확률: 1(15%), 2(25%), 3(30%), 4(10%), 5(10%), 6(10%)
dice = np.random.choice(range(1, 7), 1000000, p=[0.15, 0.25, 0.3, 0.1, 0.1, 0.1])

# 2. 그래프 설정
plt.figure(figsize=(5, 3), dpi=200)

# 3. 히스토그램 그리기
# bins=np.arange(.5, 7): [0.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5] 구간 설정
plt.hist(dice, rwidth=.98, bins=np.arange(.5, 7))

# 4. 축 설정 및 저장
plt.xticks(range(1, 7))
plt.savefig('주사위.png')
plt.show()
```

```
import matplotlib.pyplot as plt

# 도화지 크기 설정 (가로 9, 세로 4)
plt.figure(figsize=(9, 4))

# 1번 자리에 꺾은선 그래프 그리기
plt.subplot(2, 3, 1)
plt.plot([1, 2, 3])

# 5번 자리는 빈 공간으로 만들기 (subplot만 선언)
plt.subplot(2, 3, 5)

# 6번 자리에 꺾은선 그래프 그리기
plt.subplot(2, 3, 6)
plt.plot([5, 4, 3])

plt.show()
```

```
import matplotlib.pyplot as plt

# 1. 2행 4열의 서브플롯 생성
# plt.figure(figsize=(9, 4))는 subplots 사용 시 반영이 안 될 수 있어 주석 처리됨
fig, axs = plt.subplots(2, 4)

# 2. 캔버스의 크기를 인치 단위로 직접 지정
fig.set_size_inches((9, 4))

# 3. 반환받은 axs 배열을 인덱스(행, 열)로 접근하여 그림 그리기
# [0, 1] -> 1행 2열 자리
axs[0, 1].plot([5, 1, 3])

# [1, 0] -> 2행 1열 자리
axs[1, 0].plot([3, 3, 1])

# 4. plt.plot()을 직접 호출하면 마지막으로 활성화된 칸에 그려짐
plt.plot([10, 5, 20])

plt.show()
```

```
# plt.subplots()에서 직접 캔버스의 크기 지정

import matplotlib.pyplot as plt

# plt.subplots()로 figure와 크기를 지정
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

# 각 서브플롯에 데이터 추가
axes[0].plot([1, 2, 3], [1, 4, 9])
axes[1].plot([1, 2, 3], [1, 2, 3])

plt.show()
```

```
import matplotlib.pyplot as plt

# 1. 전체 도화지(figure) 생성
fig = plt.figure()

# 2. 부분 그림(subplots) 추가
# add_subplot(행, 열, 인덱스)
# 321의 의미: 3행 2열로 나눈 격자에서 1번째 칸이라는 뜻입니다.
fig.add_subplot(321).set_title('321')
fig.add_subplot(322).set_title('322')
fig.add_subplot(323).set_title('323')
fig.add_subplot(324).set_title('324')
fig.add_subplot(325).set_title('325')
fig.add_subplot(326).set_title('326')

# 3. 그래프 간격 조절
# wspace: 좌우 간격, hspace: 상하 간격
fig.subplots_adjust(wspace=0.2, hspace=0.7)

# 4. 출력
plt.show()
```

```
import matplotlib.pyplot as plt

# 1. 2행 3열의 서브플롯 생성 및 크기 지정
fig, ax = plt.subplots(2, 3, figsize=(7, 4))
fig.set_size_inches(6, 3)

# 2. 특정 칸에 데이터 그래프 그리기
ax[0, 1].plot([1, 4, 9, 16])  # (0, 1) 칸: 제곱수
ax[1, 0].plot([1, 2, 3, 4])  # (1, 0) 칸: 선형
ax[1, 2].plot([6, 5, 2, 1])  # (1, 2) 칸: 감소

# 3. 전체 제목 설정 (Figure 레벨)
fig.suptitle('전체 제목', fontsize=14, fontweight='bold')

# 4. 각 서브플롯별 제목 설정 (Axes 레벨)
ax[0, 0].set_title('(0,0) 빈 플롯')
ax[0, 1].set_title('(0,1) 제곱수')
ax[0, 2].set_title('(0,2) 빈 플롯')
ax[1, 0].set_title('(1,0) 선형')
ax[1, 1].set_title('(1,1) 빈 플롯')
ax[1, 2].set_title('(1,2) 감소')

# 5. 레이아웃 조정 및 출력
plt.tight_layout()
plt.show()
```
