```python
ax = sns.countplot(mtc, x='gear', order=mtc.gear.value_counts().index)

for p in ax.patches:
  height = p.get_height()
  ax.text(x = (p.get_x() + p.get_width() / 2) , y=(height + .2),
          s= int(height), ha = 'center',size=12,color='red')

ax.set_ylim(0,17)
plt.show()
```

```python
import matplotlib.pyplot as plt  # 그래프 그리는 라이브러리

# 예시 데이터 (mtcars의 mpg 값처럼 사용)
mtc_mpg = [21, 21, 22.8, 21.4, 18.7, 18.1, 14.3, 24.4, 22.8,
           19.2, 17.8, 16.4, 17.3, 15.2, 10.4, 10.4, 14.7,
           32.4, 30.4, 33.9, 21.5, 15.5, 15.2, 13.3, 19.2,
           27.3, 26.0, 30.4, 15.8, 19.7, 15.0, 21.4]

# 히스토그램 생성
result = plt.hist(mtc_mpg ,edgecolor='white')
# result[0] → 각 구간의 빈도수
# result[1] → 구간 경계값
# result[2] → 그래프 객체

plt.show()  # 그래프 출력

print(result)  # 그림 위에 나온 array 값 출력
```

```python
from pydataset import data  # pydataset 라이브러리에서 data 함수 가져오기

iris = data('iris')        # 'iris' 데이터셋 불러오기 (꽃 데이터)
iris.info()                # 데이터 구조 확인 (컬럼, 타입, 개수 등 출력)
```

```python
import matplotlib.pyplot as plt   # 그래프 라이브러리
from pydataset import data        # 데이터셋 가져오기

iris = data('iris')               # iris 데이터 불러오기

# 히스토그램 생성
n, bins, patches = plt.hist(
    iris['Sepal.Length'],        # 꽃받침 길이 데이터
    bins=20,                     # 구간 20개로 나눔
    color='yellowgreen',         # 막대 색상
    edgecolor='white'            # 테두리 색상
)

# 결과 출력
print(n)       # 각 구간의 빈도수
print(bins)    # 구간 경계값
print(patches) # 그래프 객체 정보

plt.show()     # 그래프 출력
```

```python
n, bins, patches = plt.hist(
    iris['Sepal.Length'],        # 꽃받침 길이 데이터
    bins=20,                     # 구간 20개로 나눔
    color='yellowgreen',         # 막대 색상
    edgecolor='white'            # 테두리 색상
)
for i in range(len(patches)):
  patches[i].set_facecolor(plt.cm.viridis(n[i]/max(n)))

plt.show()
```
