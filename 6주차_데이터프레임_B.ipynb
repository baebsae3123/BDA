```
# 1. 원본 데이터 복사
df2 = df.copy()

# 2. 새로운 열('재해자수') 추가
# '사망(명)'과 '부상(명)' 열을 가로 방향(axis=1)으로 더함
df2['재해자수'] = df[['사망(명)', '부상(명)']].sum(axis=1)
df2

```

```
# 3. 각 열의 합계 계산 (기본값 axis=0, 세로 합계)
df2.sum()

# 4. 새로운 행('총계') 추가
# 데이터프레임의 모든 열에 대해 세로 방향(axis=0) 합계를 구해 '총계'라는 이름의 행으로 저장
df2.loc['총계'] = df2.sum(axis=0)
df2
```
