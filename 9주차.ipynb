```python
from IPython.display import display, HTML

def display_side_by_side(*args):
    """데이터프레임들을 가로로 나란히 출력하는 함수 정의"""
    html_str = ''
    for df in args:
        # 데이터프레임을 HTML 표로 변환하고 스타일을 지정합니다.
        html_str += df.to_html().replace('table', 'table style="display:inline; margin:10px;"')

    display(HTML(html_str))
```

```python
display_side_by_side(corr, corr.iloc[1:, :-1])
```
